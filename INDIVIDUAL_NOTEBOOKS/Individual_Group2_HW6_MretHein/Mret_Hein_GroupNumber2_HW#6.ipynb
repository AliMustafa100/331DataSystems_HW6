# My 10 Propositions:

### Proposition 1: Employees whose OrganizationLevel is 3 and also whose HireDate is after 2008


-This query helps HR and management analyze hiring trends at a specific organizational tier.

-Understanding how many mid-level hires occurred after 2008,

-Reviewing experience mix within the organization, and

-Identifying potential candidates for leadership development or promotion.

````python
%%sql
USE AdventureWorks2019;

SELECT e.BusinessEntityID, p.FirstName, p.LastName, e.OrganizationLevel, e.HireDate
FROM HumanResources.Employee AS e
JOIN Person.Person AS p ON e.BusinessEntityID = p.BusinessEntityID
WHERE e.OrganizationLevel = 3

INTERSECT


SELECT e.BusinessEntityID, p.FirstName, p.LastName, e.OrganizationLevel, e.HireDate
FROM HumanResources.Employee AS e
JOIN Person.Person AS p ON e.BusinessEntityID = p.BusinessEntityID
WHERE e.HireDate > '2008-12-31'
ORDER BY HireDate;
````

### Proposition 2: Employees who are not salesperson


-Helps HR or management distinguish support and technical staff from sales roles.

-Useful for budgeting, assigning sales quotas, or identifying non-commissioned employees.

-Perfectly demonstrates set difference logic, a core skill from Chapter 6 of T-SQL Fundamentals.

````python
%%sql
USE AdventureWorks2019;


SELECT BusinessEntityID, JobTitle, HireDate
FROM (
    -- All employees in the company
    SELECT BusinessEntityID, JobTitle, HireDate
    FROM HumanResources.Employee

    EXCEPT
    
-- Employees who are registered as salespersons
    SELECT e.BusinessEntityID, e.JobTitle, e.HireDate
    FROM HumanResources.Employee AS e
    JOIN Sales.SalesPerson AS sp 
        ON e.BusinessEntityID = sp.BusinessEntityID
) AS NonSalesEmployees
ORDER BY BusinessEntityID;
````

### Proposition 3: People who are either salesperson or customer but not both

-Reveals exclusive roles, which can be useful for:

-CRM management: ensuring proper tagging of clients vs. staff.

-Conflict-of-interest audits: catching internal staff with unintended customer records.

-Marketing segmentation: differentiating between pure buyers and active company agents.

````python
%%sql
-- People who are customers but not salespersons
(
    SELECT TOP (20) p.BusinessEntityID, p.FirstName, p.LastName, 'Customer Only' AS RoleType
    FROM Person.Person AS p
    JOIN Sales.Customer AS c 
        ON p.BusinessEntityID = c.PersonID
    EXCEPT
    SELECT p.BusinessEntityID, p.FirstName, p.LastName, 'Customer Only' AS RoleType
    FROM Person.Person AS p
    JOIN Sales.SalesPerson AS sp 
        ON p.BusinessEntityID = sp.BusinessEntityID
)
UNION
-- People who are salespersons but not customers
(
    SELECT TOP (20) p.BusinessEntityID, p.FirstName, p.LastName, 'Salesperson Only' AS RoleType
    FROM Person.Person AS p
    JOIN Sales.SalesPerson AS sp 
        ON p.BusinessEntityID = sp.BusinessEntityID
    EXCEPT
    SELECT p.BusinessEntityID, p.FirstName, p.LastName, 'Salesperson Only' AS RoleType
    FROM Person.Person AS p
    JOIN Sales.Customer AS c 
        ON p.BusinessEntityID = c.PersonID
)
ORDER BY LastName, FirstName;
````

### Proposition 4: People who are no longer customers

-Identifies lapsed or churned customers — essential for re-engagement campaigns or retention analysis.

-Sales and marketing can reach out to these people with incentives or surveys.

````python
%%sql
-- People who have placed at least one order in the past (former customers)
SELECT DISTINCT p.BusinessEntityID, p.FirstName, p.LastName
FROM Person.Person AS p
JOIN Sales.Customer AS c 
    ON p.BusinessEntityID = c.PersonID
JOIN Sales.SalesOrderHeader AS soh 
    ON c.CustomerID = soh.CustomerID

EXCEPT

-- People who are currently listed as customers
SELECT DISTINCT p.BusinessEntityID, p.FirstName, p.LastName
FROM Person.Person AS p
JOIN Sales.Customer AS c 
    ON p.BusinessEntityID = c.PersonID
WHERE c.PersonID IS NOT NULL
ORDER BY LastName, FirstName;
````

### Proposition 5: Employees who are both female and married but the sickleavehours is more than 40


-Useful for HR analytics, identifying specific employee groups that may need policy review or wellness outreach.

````python
%%sql
-- Female employees
SELECT BusinessEntityID, JobTitle, Gender, MaritalStatus, SickLeaveHours
FROM HumanResources.Employee
WHERE Gender = 'F'

INTERSECT

-- Married employees with high sick leave
SELECT BusinessEntityID, JobTitle, Gender, MaritalStatus, SickLeaveHours
FROM HumanResources.Employee
WHERE MaritalStatus = 'M'
  AND SickLeaveHours > 40
ORDER BY BusinessEntityID;
````

### Proposition 6: Products which listprice history is between 2011 and 2013

-Helps finance and product management teams analyze pricing trends across a specific time period.

-Can reveal how many products were actively repriced during a given market cycle (e.g., inflation response).

````python
%%sql
-- Products that had price history starting after 2011
SELECT DISTINCT ProductID, ListPrice, ModifiedDate
FROM Production.ProductListPriceHistory
WHERE YEAR(ModifiedDate) >= 2011

INTERSECT

-- Products that had price history before or during 2013
SELECT DISTINCT ProductID, ListPrice, ModifiedDate
FROM Production.ProductListPriceHistory
WHERE YEAR(ModifiedDate) <= 2013
ORDER BY ProductID;
````

### Proposition 7:  Products which are both clothing and accessories


-Useful for inventory audits and data quality checks — finding products mistakenly tagged in multiple categories.

-Helps marketing and merchandising teams spot cross-category items that may need unique campaigns.

````python
%%sql
-- All products that belong to the Clothing category
SELECT p.ProductID, p.Name AS ProductName
FROM Production.Product AS p
JOIN Production.ProductSubcategory AS ps
    ON p.ProductSubcategoryID = ps.ProductSubcategoryID
JOIN Production.ProductCategory AS pc
    ON ps.ProductCategoryID = pc.ProductCategoryID
WHERE pc.Name = 'Clothing'

INTERSECT

-- All products that belong to the Accessories category
SELECT p.ProductID, p.Name AS ProductName
FROM Production.Product AS p
JOIN Production.ProductSubcategory AS ps
    ON p.ProductSubcategoryID = ps.ProductSubcategoryID
JOIN Production.ProductCategory AS pc
    ON ps.ProductCategoryID = pc.ProductCategoryID
WHERE pc.Name = 'Accessories'
ORDER BY ProductName;
````

### Proposition 8: Products Sold in 2012 but Not in 2013


-Detects product churn or items that stopped selling year-over-year.

-Helps marketing decide which SKUs need remarketing or retirement.

````python
%%sql
-- Products that were sold in 2012
SELECT DISTINCT p.ProductID, p.Name AS ProductName
FROM Production.Product AS p
JOIN Sales.SalesOrderDetail AS sod ON p.ProductID = sod.ProductID
JOIN Sales.SalesOrderHeader AS soh ON sod.SalesOrderID = soh.SalesOrderID
WHERE YEAR(soh.OrderDate) = 2012

EXCEPT

-- Products that were sold in 2013
SELECT DISTINCT p.ProductID, p.Name AS ProductName
FROM Production.Product AS p
JOIN Sales.SalesOrderDetail AS sod ON p.ProductID = sod.ProductID
JOIN Sales.SalesOrderHeader AS soh ON sod.SalesOrderID = soh.SalesOrderID
WHERE YEAR(soh.OrderDate) = 2013
ORDER BY ProductName;
````

### Proposition 9: Employees in Sales or Research Departments (But Not Both)


-Reveals cross-functional overlap or separation of roles.

-Useful for HR when checking department assignments or reorg planning.

````python
%%sql
(
-- Employees in Research and Development but not sales
    SELECT e.BusinessEntityID, e.JobTitle, d.Name AS Department
    FROM HumanResources.Employee AS e
    JOIN HumanResources.EmployeeDepartmentHistory AS edh ON e.BusinessEntityID = edh.BusinessEntityID
    JOIN HumanResources.Department AS d ON edh.DepartmentID = d.DepartmentID
    WHERE d.Name = 'Research and Development'

    EXCEPT

    -- Employees in Sales
    SELECT e.BusinessEntityID, e.JobTitle, d.Name AS Department
    FROM HumanResources.Employee AS e
    JOIN HumanResources.EmployeeDepartmentHistory AS edh ON e.BusinessEntityID = edh.BusinessEntityID
    JOIN HumanResources.Department AS d ON edh.DepartmentID = d.DepartmentID
    WHERE d.Name = 'Sales'
)
UNION
(
    -- Employees in Sales but not Research and Development
    SELECT e.BusinessEntityID, e.JobTitle, d.Name AS Department
    FROM HumanResources.Employee AS e
    JOIN HumanResources.EmployeeDepartmentHistory AS edh ON e.BusinessEntityID = edh.BusinessEntityID
    JOIN HumanResources.Department AS d ON edh.DepartmentID = d.DepartmentID
    WHERE d.Name = 'Sales'

    EXCEPT

    SELECT e.BusinessEntityID, e.JobTitle, d.Name AS Department
    FROM HumanResources.Employee AS e
    JOIN HumanResources.EmployeeDepartmentHistory AS edh ON e.BusinessEntityID = edh.BusinessEntityID
    JOIN HumanResources.Department AS d ON edh.DepartmentID = d.DepartmentID
    WHERE d.Name = 'Research and Development'
)
ORDER BY BusinessEntityID;
````

### Proposition 10: Products with Low Safety Stock, High Reorder Point, and Non-Zero Price



-Highlights products that may be at risk of stockouts but are actively sold and valuable.

-Supports supply-chain and purchasing teams in setting smarter restock thresholds.

-Demonstrates a real-world use of set intersection logic to combine multiple conditions cleanly,

````python
%%sql
-- Products with safety stock level less than 1000
SELECT ProductID, Name, SafetyStockLevel, ReorderPoint, ListPrice
FROM Production.Product
WHERE SafetyStockLevel < 1000

INTERSECT

-- Products with reorder point greater than 300
SELECT ProductID, Name, SafetyStockLevel, ReorderPoint, ListPrice
FROM Production.Product
WHERE ReorderPoint > 300

INTERSECT

-- Products with list price not equal to 0
SELECT ProductID, Name, SafetyStockLevel, ReorderPoint, ListPrice
FROM Production.Product
WHERE ListPrice <> 0
ORDER BY ProductID;
````
